```python
%matplotlib inline
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import norm
import pandas
```

```python
df = pandas.read_csv("data/TCL_wt1.tsv", sep="\t", na_values="#VALEUR!")
df = df.dropna()

v = df['Log2 Corrected Abundance Ratio'].tolist()

fig, ax = plt.subplots()
ax.hist(v, bins=50)

ax.set_title("Histogramme des Log2 Corrected Abundance Ratio")
ax.set_xlabel("Log2 Corrected Abundance Ratio")
ax.set_ylabel("Nombre de protéines")

fig.show()
```

```python
fig.savefig("histogram.png")
```
