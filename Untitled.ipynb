```python
surabondantes = df[(df['Log2 Corrected Abundance Ratio'] > mean + et) & 
                   (df['-LOG10 Adj.P-val'] > 3)]

uniprot_ids = surabondantes['Accession']
uniprot_ids.tolist()
```

```python
import json
from scipy.stats import hypergeom

with open("data/EColiK12_GOcounts.json", "r") as fp:
    background = json.load(fp)["go_terms"]
```

```python
go_dico = {}

for acc in uniprot_ids:
    go_terms = getAccessionGOTerms("data/uniprot-proteome_UP000000625.xml", acc)
    
    for go_id, go_name in go_terms:
        
        if go_id not in go_dico:
            go_dico[go_id] = {
                "ID": go_id,
                "name": go_name,
                "carried_by": []
            }
        
        go_dico[go_id]["carried_by"].append(acc)
```

```python
def compute_pvalue(GO_term, background, n, N=1800):
    GO_id = GO_term["ID"]
    GO_name = GO_term["name"]
    
    k = len(GO_term["carried_by"])      # nb observé de protéines surabondantes portant ce GO
    K = background[GO_id]["count"]      # nb total de protéines du protéome portant ce GO
    
    rv = hypergeom(N, K, n)
    p_value = rv.sf(k - 1)              # P(X >= k)
    
    return {
        "GO_ID": GO_id,
        "definition": GO_name,
        "occurrence": k,
        "pvalue": p_value
    }
```

```python
n = len(uniprot_ids)   # nombre total de protéines surabondantes
N = 1800               # nombre total de protéines du protéome
```

```python
results = []

for go_id in go_dico:
    results.append(compute_pvalue(go_dico[go_id], background, n, N))

results = sorted(results, key=lambda x: x["pvalue"])
```
